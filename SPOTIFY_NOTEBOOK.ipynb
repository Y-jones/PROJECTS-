```python
import pandas as pd
data_frame = pd.read_csv('spotify_tracks.csv')

long_tracks = data_frame[data_frame['popularity'] > 80]
print("------------- SONGS WITH POPULARITY > 80 ----------------")
display(long_tracks.head(10))

print("------------------------- EXPLICIT SONGS ----------------------")
display(data_frame[data_frame['explicit'] == True])

genres = data_frame['genre'].unique()   
choice = input("Which genre do you want?: ")
while(choice not in genres) :
    print(f"Available genres are {list(genres)}")
    choice = input("Genre is not available in the track, enter another one: ")
else :
    print(f"------------------------ 10 Songs of {choice} ----------------------------")
    display(data_frame[data_frame['genre'] == choice].head(10))

print("--------------------- 10 SONGS LONGER THAN AVERAGE DURATION ---------------------")
display(data_frame[data_frame['duration_ms'] > data_frame['duration_ms'].mean()].head(10))
```

```python
import pandas as pd

data_frame = pd.read_csv('spotify_tracks.csv')

genre_popularity = data_frame.groupby('genre')['popularity'].mean()

genre_popularity = genre_popularity.sort_values(ascending=False)

genre_popularity_df = genre_popularity.reset_index()

print("------------------ AVERAGE POPULARITY BY GENRE ------------------")
display(genre_popularity_df)


genre_count = data_frame.groupby('genre')['name'].count()
genre_count = genre_count.sort_values(ascending=False)
genre_count_dataframe = genre_count.reset_index()
print("----------------------- NUMBER OF SONGS BY GENRE --------------------")
display(genre_count_dataframe)

avrg_duration = data_frame.groupby('genre')['duration_ms'].mean()
avrg_duration = avrg_duration.sort_values(ascending=False)
avrg_duration_dataframe = avrg_duration.reset_index()
print("-------------------------- AVERAGE DURATION BY GENRE ----------------------------")
display(avrg_duration_dataframe)

top_songs = data_frame.groupby('genre')['popularity'].mean()
top_songs = top_songs.sort_values(ascending=False)
top_songs_datafame = top_songs.reset_index()
print("----------------------------- TOP GENRES BY POPULARITY -------------------------")
display(top_songs_datafame)
print(data_frame.columns)

```

```python
import pandas as pd
import numpy as np

data_frame = pd.read_csv("spotify_tracks.csv")

conditions = [
    data_frame["genre"].str.contains(
        "rock|grunge|guitar|metal|punk|emo", case=False, na=False
    ),
    data_frame["genre"].str.contains("pop|indie|j-idol|k-pop", case=False, na=False),
    data_frame["genre"].str.contains(
        "house|techno|dance|edm|electro|trance|disco|club|breakbeat|drum-and-bass|dubstep",
        case=False,
        na=False,
    ),
    data_frame["genre"].str.contains(
        "hip-hop|rap|r-n-b|soul|funk|groove", case=False, na=False
    ),
    data_frame["genre"].str.contains(
        "reggae|reggaeton|salsa|samba|tango|latin|latino", case=False, na=False
    ),
    data_frame["genre"].str.contains(
        "ambient|chill|sleep|study|rainy-day|piano|acoustic|classical|opera|new-age",
        case=False,
        na=False,
    ),
    data_frame["genre"].str.contains(
        "jazz|blues|gospel|folk|country|bluegrass|honky-tonk",
        case=False,
        na=False,
    ),
    data_frame["genre"].str.contains(
        "disney|children|kids|comedy|show-tunes|movies|soundtracks",
        case=False,
        na=False,
    ),
]

categories = [
    "Rock/Metal",
    "Pop/Indie",
    "Electronic/Dance",
    "Hip-Hop/R&B",
    "Latin/Reggae",
    "Ambient/Chill/Classical",
    "Jazz/Blues/Folk/Country",
    "Media/Kids/Comedy",
]

data_frame["genre_category"] = np.select(conditions, categories, default="Other/World")

data_frame.to_csv("spotify_tracks_minutes_merged.csv")
```
